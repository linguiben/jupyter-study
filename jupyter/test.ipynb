````python
import pandas  as pd
obj = pd.Series(['1', '2', '1', '2', '3', '1', '2', '2', '3']) 
# obj = pd.Series(['c', 'a', 'd', 'a', 'a', 'b', 'b', 'c', 'c'])
# print(obj)
# obj.value_counts()
obj.value_counts(sort=False)
print("======分割线======")
data = pd.DataFrame({'A': [1, 3, 4, 3, 4],
                     'B': [2, 3, 1, 2, 3],
                     'C': [1, 5, 2, 4, 4]}, index=['x', 'y', 'z', 'q', 't']) # ABC表示列名，index表示行名
print(f"data 的值是: \n{data}")
# 取data的行和列
data['A']  # 取A列
data.loc['x']  # 取x行
data.iloc[0]  # 取第0行
print(f"取A列和B列，返回DataFrame:\n{data.loc[:, ['A', 'B']]}") # 取A列和B列，返回类型是DataFrame
print(f"取x行的A列，返回Series:\n{data.loc['x', ['A','B']]}") # 取x行的A列和B列，返回类型是Series
data.loc[['x', 'y'], ['A', 'B']] # 取x行和y行的A列和B列
print(data.apply(pd.value_counts)) # 统计每一列的频数，表示每一列每个值出现的次数，返回结果是DataFrame


#result = data.apply(pd.value_counts) 

print("======分组=====")
data.groupby('A')
for name, group in data.groupby('A'):
    print(name)
    print(group)
print("======统计======")
result = data.apply(pd.value_counts).fillna(0)   #在列各数字出现多少次
print(result)
print("======分割线======")
````

````python
status = 418
def http_error(status):
    match status:
        case 400:
            return "Bad request"
        case 401 | 403 | 404:
            return "Not found"
        case 418:
            return "I'm a teapot"
        case _:
            return "Something's wrong with the internet"
http_error(status) # 只执行匹配的case
````
